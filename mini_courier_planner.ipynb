# Mini-Courier Planner
Run from the project root. This notebook uses the tested Python modules and sample datasets.

```python
from main import load_parcels, load_graph
from src.algorithms import prioritize_parcels, knapsack_select, dijkstra, nearest_neighbor_tour
from src.benchmark import run_benchmarks
```

## 1. Load sample datasets

```python
parcels = load_parcels()
parcels = [{**p, 'deadline_hours': int(p['deadline_hours']), 'priority': int(p['priority']), 'weight': int(p['weight'])} for p in parcels]
graph, depot = load_graph()
print('Parcels:', len(parcels), '| Locations:', list(graph))
```

## 2. Sort parcels

```python
ordered = prioritize_parcels(parcels)
for parcel in ordered: print(parcel)
```

## 3. 0/1 Knapsack packing plan

```python
selected, total_priority, dp_table = knapsack_select(ordered, capacity=18)
print('Selected IDs:', [p['id'] for p in selected])
print('Total weight:', sum(p['weight'] for p in selected), 'kg')
print('Total priority:', total_priority)
```

## 4. Dijkstra shortest paths

```python
distances, previous = dijkstra(graph, depot)
for node, distance in distances.items(): print(f'{depot} -> {node}: {distance:.1f} km')
```

## 5. Nearest-neighbor delivery tour

```python
tour, total_distance = nearest_neighbor_tour(graph, depot)
print(' -> '.join(tour))
print(f'Tour distance: {total_distance:.1f} km')
```

## 6. Runtime benchmark

```python
for result in run_benchmarks(): print(result)
```

## 7. Complexity summary
- Sorting: O(n log n)
- 0/1 Knapsack: O(nW) time and O(nW) table space
- Dijkstra with a binary heap: O((V+E) log V)
- Nearest neighbor: repeated shortest-path computations; heuristic, not guaranteed optimal

Sample data is synthetic. Timings vary by computer.
